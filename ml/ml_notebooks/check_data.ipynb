```python
import requests
import pandas as pd
from getpass import getpass

URL = 'https://web.tk-ekat.ru'
login = getpass('Login')
password = getpass('Password')
```

```python
def to_login(url: str, login: str, password: str):
    url += '/ServiceJSON/Login'
    data = {
        'UserName': login,
        'Password': password
    }

    return requests.get(url, data).content

def get_schemas(url: str, session: str):
    url += '/ServiceJSON/EnumSchemas'
    data = {
        'session': session
    }

    response = requests.get(url, data)
    return response.content
```

```python
SESSION = to_login(URL, login, password)
```

```python
import json

schemas = get_schemas(URL, SESSION)
schemas = json.loads(schemas)

schemas
```
